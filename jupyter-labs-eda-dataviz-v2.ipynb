```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print('Libraries imported')
```

## Load Dataset

```python
df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_2.csv')
df.head(5)
```

```python
print('Shape:', df.shape)
print('Columns:', df.columns.tolist())
```

## Visualize FlightNumber vs PayloadMass

```python
sns.catplot(y='PayloadMass', x='FlightNumber', hue='Class', data=df, aspect=5)
plt.xlabel('Flight Number', fontsize=20)
plt.ylabel('Pay load Mass (kg)', fontsize=20)
plt.title('Flight Number vs Payload Mass by Landing Class')
plt.show()
```

## TASK 1: Visualize Flight Number vs Launch Site

```python
sns.catplot(y='LaunchSite', x='FlightNumber', hue='Class', data=df, aspect=1)
plt.ylabel('Launch Site', fontsize=15)
plt.xlabel('Flight Number', fontsize=15)
plt.title('Flight Number vs Launch Site')
plt.show()
print('Insight: As flight number increases, landing success rate improves. KSC LC-39A shows consistent success in later flights.')
```

## TASK 2: Visualize Payload vs Launch Site

```python
sns.catplot(y='LaunchSite', x='PayloadMass', hue='Class', data=df, aspect=1)
plt.ylabel('Launch Site', fontsize=15)
plt.xlabel('Payload Mass (kg)', fontsize=15)
plt.title('Payload Mass vs Launch Site')
plt.show()
print('Insight: VAFB SLC-4E has no launches with heavy payload (>10000 kg).')
```

## TASK 3: Visualize success rate of each orbit type

```python
orbit_success = df.groupby('Orbit')['Class'].mean().reset_index()
orbit_success.columns = ['Orbit', 'SuccessRate']
orbit_success = orbit_success.sort_values('SuccessRate', ascending=False)

plt.figure(figsize=(12,5))
sns.barplot(x='Orbit', y='SuccessRate', data=orbit_success, palette='viridis')
plt.xlabel('Orbit Type', fontsize=14)
plt.ylabel('Success Rate', fontsize=14)
plt.title('Landing Success Rate by Orbit Type')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
print('Insight: ES-L1, GEO, HEO, SSO orbits have 100% success rate. GTO has the lowest success rate.')
```

## TASK 4: Visualize FlightNumber vs Orbit type

```python
sns.catplot(y='Orbit', x='FlightNumber', hue='Class', data=df, aspect=1.5)
plt.ylabel('Orbit', fontsize=15)
plt.xlabel('Flight Number', fontsize=15)
plt.title('Flight Number vs Orbit')
plt.show()
print('Insight: LEO orbit success is related to number of flights. GTO shows no clear trend.')
```

## TASK 5: Visualize Payload vs Orbit type

```python
sns.catplot(y='Orbit', x='PayloadMass', hue='Class', data=df, aspect=1.5)
plt.ylabel('Orbit', fontsize=15)
plt.xlabel('Payload Mass (kg)', fontsize=15)
plt.title('Payload Mass vs Orbit')
plt.show()
```

## TASK 6: Visualize launch success yearly trend

```python
year = []
def Extract_year(date):
    for i in df['Date']:
        year.append(i.split('-')[0])
    return year

year = Extract_year(df['Date'])
df2 = df.copy()
df2['Year'] = year

year_success = df2.groupby('Year')['Class'].mean().reset_index()
year_success.columns = ['Year', 'SuccessRate']

plt.figure(figsize=(10,5))
plt.plot(year_success['Year'], year_success['SuccessRate'], marker='o', color='cyan', linewidth=2)
plt.xlabel('Year', fontsize=14)
plt.ylabel('Average Success Rate', fontsize=14)
plt.title('Launch Success Rate Trend by Year')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
print('Insight: Success rate increased significantly from 2013 and became stable/high after 2017.')
```

## Feature Engineering

```python
features = df[['FlightNumber','PayloadMass','Orbit','LaunchSite','Flights','GridFins','Reused','Legs','LandingPad','Block','ReusedCount','Serial']]
features.head()
```

## TASK 7: Create dummy variables

```python
features_one_hot = pd.get_dummies(features, columns=['Orbit','LaunchSite','LandingPad','Serial'])
print('Shape after one-hot encoding:', features_one_hot.shape)
features_one_hot.head()
```

## TASK 8: Cast all numeric columns to float64

```python
features_one_hot = features_one_hot.astype('float64')
print('Data types after casting:')
print(features_one_hot.dtypes.unique())
```
