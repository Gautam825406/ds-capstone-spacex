## Load Dataset

```python
df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv')
df.head(10)
```

## Identify and calculate percentage of missing values

```python
missing_pct = df.isnull().sum() / len(df) * 100
print('Missing value percentages:')
print(missing_pct)
```

```python
# Identify numerical and categorical columns
print('Data types:')
print(df.dtypes)
```

## TASK 1: Calculate the number of launches on each site

```python
launch_sites = df['LaunchSite'].value_counts()
print('Launches per site:')
print(launch_sites)
```

## TASK 2: Calculate number and occurrence of each orbit

```python
orbit_counts = df['Orbit'].value_counts()
print('Orbit value counts:')
print(orbit_counts)
```

## TASK 3: Calculate number and occurrence of mission outcome of the orbits

```python
landing_outcomes = df['Outcome'].value_counts()
print('Landing outcomes:')
print(landing_outcomes)
```

```python
for i, outcome in enumerate(landing_outcomes.keys()):
    print(i, outcome)
```

```python
bad_outcomes = set(landing_outcomes.keys()[[1,3,5,6,7]])
print('Bad outcomes set:')
print(bad_outcomes)
```

## TASK 4: Create a landing outcome label from Outcome column

```python
landing_class = [0 if outcome in bad_outcomes else 1 for outcome in df['Outcome']]
df['Class'] = landing_class
print('Class distribution:')
print(df['Class'].value_counts())
print('\nSuccess rate:', df['Class'].mean())
```
