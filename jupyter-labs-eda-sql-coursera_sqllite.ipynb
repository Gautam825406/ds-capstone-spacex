```python
import csv, sqlite3
import pandas as pd
import numpy as np

con = sqlite3.connect('my_data1.db')
cur = con.cursor()
print('Database connection established')
```

```python
%sql sqlite:///my_data1.db
```

```python
df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/Spacex.csv')
df.to_sql('SPACEXTBL', con, if_exists='replace', index=False, method='multi')
print('Data loaded into SQLite. Shape:', df.shape)
df.head()
```

## Task 1: Display the names of the unique launch sites

```python
%sql SELECT DISTINCT "Launch Site" FROM SPACEXTBL;
```

## Task 2: Display 5 records where launch sites begin with 'CCA'

```python
%sql SELECT * FROM SPACEXTBL WHERE "Launch Site" LIKE 'CCA%' LIMIT 5;
```

## Task 3: Display total payload mass by NASA (CRS)

```python
%sql SELECT SUM("PAYLOAD_MASS__KG_") AS Total_Payload_Mass FROM SPACEXTBL WHERE Customer = 'NASA (CRS)';
```

## Task 4: Display average payload mass by booster version F9 v1.1

```python
%sql SELECT AVG("PAYLOAD_MASS__KG_") AS Avg_Payload FROM SPACEXTBL WHERE "Booster Version" LIKE 'F9 v1.1%';
```

## Task 5: List the date of first successful landing outcome on ground pad

```python
%sql SELECT MIN(Date) AS First_Ground_Pad_Success FROM SPACEXTBL WHERE "Landing Outcome" = 'Success (ground pad)';
```

## Task 6: List the names of the boosters that landed successfully on drone ship with payload between 4000 and 6000 kg

```python
%sql SELECT "Booster Version" FROM SPACEXTBL WHERE "Landing Outcome" = 'Success (drone ship)' AND "PAYLOAD_MASS__KG_" >= 4000 AND "PAYLOAD_MASS__KG_" <= 6000;
```

## Task 7: List the total number of successful and failure mission outcomes

```python
%sql SELECT "Mission Outcome", COUNT(*) AS Count FROM SPACEXTBL GROUP BY "Mission Outcome";
```

## Task 8: List the booster versions with maximum payload mass

```python
%sql SELECT "Booster Version", "PAYLOAD_MASS__KG_" FROM SPACEXTBL WHERE "PAYLOAD_MASS__KG_" = (SELECT MAX("PAYLOAD_MASS__KG_") FROM SPACEXTBL);
```

## Task 9: List records with month=5, landing outcomes=Failure (drone ship), booster versions F9 v1.1 and F9 FT

```python
%sql SELECT * FROM SPACEXTBL WHERE strftime('%m', Date) = '05' AND "Landing Outcome" = 'Failure (drone ship)' AND ("Booster Version" LIKE 'F9 v1.1%' OR "Booster Version" LIKE 'F9 FT%');
```

## Task 10: Rank landing outcomes between 2010-06-04 and 2017-03-20 descending

```python
%sql SELECT "Landing Outcome", COUNT(*) AS Count FROM SPACEXTBL WHERE Date BETWEEN '2010-06-04' AND '2017-03-20' GROUP BY "Landing Outcome" ORDER BY Count DESC;
```
